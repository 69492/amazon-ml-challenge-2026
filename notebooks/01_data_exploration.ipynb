```python
from pathlib import Path
import pandas as pd

# Find project root by looking for student_resource
current = Path.cwd()

if (current / "student_resource").exists():
    PROJECT_ROOT = current
elif (current.parent / "student_resource").exists():
    PROJECT_ROOT = current.parent
else:
    raise FileNotFoundError(
        "Could not find student_resource folder. "
        f"Checked: {current} and {current.parent}"
    )

DATA_ROOT = PROJECT_ROOT / "student_resource"

print("Project root:", PROJECT_ROOT)
print("Student resource:", DATA_ROOT)
print("Exists:", DATA_ROOT.exists())
```

```python
tsv_files = sorted(DATA_ROOT.rglob("*.tsv"))

print(f"Found {len(tsv_files)} TSV files:\n")

for file in tsv_files:
    print(file.relative_to(DATA_ROOT))
```

```python
for file in tsv_files:
    print("\n" + "=" * 80)
    print("FILE:", file.relative_to(DATA_ROOT))

    df_sample = pd.read_csv(
        file,
        sep="\t",
        nrows=5
    )

    print("Columns:", list(df_sample.columns))
    print("Sample shape:", df_sample.shape)
    display(df_sample.head())
```

```python
source_files = {}
ground_truth_file = None

for file in tsv_files:
    columns = set(
        pd.read_csv(file, sep="\t", nrows=0).columns
    )

    if {
        "entity_id",
        "business_name",
        "business_address",
        "country"
    }.issubset(columns):
        source_files[file.name] = file

    elif {
        "source1_entity_id",
        "matched_entity_ids"
    }.issubset(columns):
        ground_truth_file = file

print("SOURCE FILES")
for name, path in source_files.items():
    print(name, "->", path)

print("\nGROUND TRUTH")
print(ground_truth_file)
```

```python
for name, path in source_files.items():
    print(name, "->", path.parent.relative_to(DATA_ROOT))
```

```python
sources = {}

for name, path in source_files.items():
    print(f"Loading {name} ...")

    sources[name] = pd.read_csv(
        path,
        sep="\t"
    )

    print("  Shape:", sources[name].shape)

print("\nLoaded datasets:", len(sources))
```

```python
for name, df in sources.items():

    print("\n" + "=" * 80)
    print(name)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    print("\nColumns:")
    print(list(df.columns))

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nDuplicate rows:", df.duplicated().sum())
```

```python
ground_truth = pd.read_csv(
    ground_truth_file,
    sep="\t"
)

print("Ground truth shape:", ground_truth.shape)

display(ground_truth.head(10))
```

```python
print(ground_truth.columns.tolist())
```

```python
def count_matches(value):
    if pd.isna(value) or str(value).strip() == "":
        return 0

    return len(
        [
            x.strip()
            for x in str(value).split(",")
            if x.strip()
        ]
    )


ground_truth["match_count"] = (
    ground_truth["matched_entity_ids"]
    .apply(count_matches)
)

print(
    ground_truth["match_count"]
    .value_counts()
    .sort_index()
)
```

```python
print("\nMatch distribution:")

print(
    ground_truth["match_count"]
    .value_counts()
    .sort_index()
)

print("\nPercentage:")

print(
    (
        ground_truth["match_count"]
        .value_counts(normalize=True)
        .sort_index()
        * 100
    ).round(2)
)
```
