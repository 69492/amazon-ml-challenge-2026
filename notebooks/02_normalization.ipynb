```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\konda\23BQ1A4223\amazon-ml-challenge-2026\student_resource\dataset\train\train_source1.tsv",
    sep="\t"
)

df = add_normalized_columns(df)

display(
    df[
        [
            "business_name",
            "business_name_normalized",
            "business_address",
            "business_address_normalized",
            "country",
            "country_normalized",
        ]
    ].head(20)
)
```

```python
df = add_normalized_columns(df)

print("Columns after normalization:")
print(df.columns.tolist())
```

```python
display(
    df[
        [
            "business_name",
            "business_name_normalized",
            "business_address",
            "business_address_normalized",
            "country",
            "country_normalized",
        ]
    ].head(20)
)
```

```python
original_name = df["business_name"].fillna("").astype(str)
normalized_name = df["business_name_normalized"]

original_address = df["business_address"].fillna("").astype(str)
normalized_address = df["business_address_normalized"]

name_changed = original_name != normalized_name
address_changed = original_address != normalized_address

print(
    "Business names changed:",
    name_changed.sum(),
    "/",
    len(df)
)

print(
    "Addresses changed:",
    address_changed.sum(),
    "/",
    len(df)
)
```

```python
display(
    df.loc[
        name_changed,
        [
            "business_name",
            "business_name_normalized"
        ]
    ].head(30)
)
```

```python
display(
    df.loc[
        address_changed,
        [
            "business_address",
            "business_address_normalized"
        ]
    ].head(30)
)
```

```python
print("Empty original business names:",
      (df["business_name"].fillna("").astype(str).str.strip() == "").sum())

print("Empty normalized business names:",
      (df["business_name_normalized"].str.strip() == "").sum())

print()

print("Empty original addresses:",
      (df["business_address"].fillna("").astype(str).str.strip() == "").sum())

print("Empty normalized addresses:",
      (df["business_address_normalized"].str.strip() == "").sum())
```

```python
print("Original countries:")
print(df["country"].value_counts(dropna=False))

print("\nNormalized countries:")
print(df["country_normalized"].value_counts(dropna=False))
```

```python
name_counts = (
    df["business_name_normalized"]
    .value_counts()
)

print("Unique normalized business names:",
      name_counts.size)

print("Total rows:",
      len(df))

print("\nMost common normalized names:")

display(
    name_counts
    .head(20)
    .to_frame("count")
)
```

```python
unique_name_rows = (
    df["business_name_normalized"]
    .value_counts()
)

print(
    "Names appearing exactly once:",
    (unique_name_rows == 1).sum()
)

print(
    "Names appearing more than once:",
    (unique_name_rows > 1).sum()
)

print(
    "Maximum frequency of one normalized name:",
    unique_name_rows.max()
)
```
